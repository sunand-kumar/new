```python
# -----------------------------
# 📘 AI Question Paper Generator using Groq
# With PDF Export
# -----------------------------

from groq import Groq
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas
import textwrap

# 🔑 Paste your Groq API Key
client = Groq(
    api_key="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"
)

# -----------------------------
# Function: Generate Question Paper
# -----------------------------
def generate_question_paper(subject, topics, num_questions=5, difficulty="medium", q_type="short answer"):
    prompt = f"""
    You are an expert examiner.
    Create a question paper with the following requirements:
    - Subject: {subject}
    - Topics: {topics}
    - Number of Questions: {num_questions}
    - Difficulty Level: {difficulty}
    - Question Type: {q_type}

    Rules:
    1. Do not repeat questions.
    2. Keep the questions clear and well-structured.
    3. Output must be formatted as:
       Q1. ...
       Q2. ...
       etc.
    """

    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "system", "content": "You are a strict and detail-oriented exam paper setter."},
            {"role": "user", "content": prompt}
        ],
        temperature=0.7
    )

    return response.choices[0].message.content


# -----------------------------
# Function: Save Questions to PDF
# -----------------------------
def save_to_pdf(text, filename="Question_Paper.pdf"):
    c = canvas.Canvas(filename, pagesize=A4)
    width, height = A4

    # Title
    c.setFont("Helvetica-Bold", 16)
    c.drawString(200, height - 50, "Generated Question Paper")

    # Body text
    c.setFont("Helvetica", 12)
    y = height - 100
    for line in text.split("\n"):
        wrapped = textwrap.wrap(line, width=90)
        for wrap_line in wrapped:
            c.drawString(50, y, wrap_line)
            y -= 20
            if y < 50:  # New page if space runs out
                c.showPage()
                c.setFont("Helvetica", 12)
                y = height - 50
    c.save()
    print(f"✅ Question Paper saved as {filename}")


# -----------------------------
# Example Run
# -----------------------------
subject = "Mathematics"
topics = "Algebra, Quadratic Equations"
questions = generate_question_paper(subject, topics, num_questions=5, difficulty="hard", q_type="MCQ")

print("📄 Generated Question Paper:\n")
print(questions)

# Save to PDF
save_to_pdf(questions, "Math_Question_Paper.pdf")
```
